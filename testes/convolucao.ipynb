```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#sharpness
im = cv2.imread('dog.jpg')

im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
```

```python
def plot_imgs_inline(imgs):
    size = len(imgs)
    for i in range(size):
        plt.subplot(1,size,i+1),plt.imshow(imgs[i])
    plt.show()
    
def plot_imgs(imgs):
    size = len(imgs)
    for i in range(size):
        plt.imshow(imgs[i])
        plt.show()
```
