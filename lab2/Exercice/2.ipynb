```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from cv_utils import waitKey
```

```python
# Carregar as imagens
circle_img = cv2.imread('./imagens/circle.jpg')
line_img = cv2.imread('./imagens/line.jpg')

# cv2.imshow('threashold',line_img)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
height, width = 300, 300
background = np.full((height, width), 255, dtype=np.uint8)
```

```python
# Função para rotacionar imagens
def rotate_image(img, angle):
    width = img.shape[1]
    height = img.shape[0]
    M_rotation = cv2.getRotationMatrix2D((width / 2, height / 2), angle, 1)
    rotated = cv2.warpAffine(img, M_rotation, (width, height), borderValue=255)
    return rotated

# Função para redimensionar imagens
def resize_image(im, new_width, new_height):
    return cv2.resize(im, (new_width, new_height))

# Função para colar as imagens com máscara
def paste_image(background, img, x, y):
    h, w,_ = img.shape
    # Criar uma máscara binária onde a parte não branca é 1 e o fundo é 0
    mask = img < 255
    # Inserir a imagem no fundo onde a máscara for verdadeira
    background[y:y+h, x:x+w][mask] = img[mask]
```

```python
# Criando a cabeça do boneco (círculo)
circle_resized = resize_image(circle_img, 50, 50)  # Definindo a cabeça como 50x50
paste_image(background, circle_resized, 125, 50)  # Posicionar a cabeça no topo

# Criando o tronco (linha)
line_resized_tronco = resize_image(line_img, 10, 100)  # Tronco de 100 pixels de altura
paste_image(background, line_resized_tronco, 145, 100)  # Posicionar tronco logo abaixo da cabeça

# Criando os braços (75% do tronco)
line_resized_braco = resize_image(line_img, 10, 75)
arm_left = rotate_image(line_resized_braco, -45)  # Braço esquerdo rotacionado a -45 graus
arm_right = rotate_image(line_resized_braco, 45)  # Braço direito rotacionado a 45 graus
paste_image(background, arm_left, 70, 125)  # Posicionar braço esquerdo
paste_image(background, arm_right, 155, 125)  # Posicionar braço direito

# Criando as pernas (duplo do tamanho dos braços, em 90º entre si)
line_resized_perna = resize_image(line_img, 10, 150)
leg_left = rotate_image(line_resized_perna, 90)  # Perna esquerda a 90 graus
leg_right = rotate_image(line_resized_perna, 45)  # Perna direita a 45 graus
paste_image(background, leg_left, 90, 200)  # Posicionar perna esquerda
paste_image(background, leg_right, 150, 200)  # Posicionar perna direita

# Mostrar a imagem final
cv2.imshow('Boneco Palito', background)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
