```python
# Carregar as imagens
circle_img = cv2.imread('./imagens/circle.jpg')
line_img = cv2.imread('./imagens/line.jpg')
```

```python
height, width = 300, 300
background = np.full((height, width),255, dtype=np.uint8)
```

```python
#retorna a imagem rotacionada
def rotate_image(img, angle):
    width = img.shape[1]
    height = img.shape[0]
    x_center = width/2
    y_center = height/2
    
    M_rotation = cv2.getRotationMatrix2D((x_center,y_center),angle,1)
    im_rotated = cv2.warpAffine(img,M_rotation,(width, height))
    return im_rotated

def translate_image(img, x=0, y=0):
    width = img.shape[1]
    height = img.shape[0]
    M_translation = np.float32([[1,0,x],[0,1,y]])
    im_translated = cv2.warpAffine(img,M_translation,(width,height))
    return im_translated
    
def scaling_image(img, xScale = 1, yScale = 1 ):
    height, width = img.shape[:2]
    
    new_height = int(height * yScale)
    new_width = int(width * xScale)
    x_offset = abs(new_width - width) // 2
    y_offset = abs(new_height - height) // 2
    
    M_scaling = np.float32([[xScale, 0, x_offset], [0, yScale, y_offset]])

    im_scaled = cv2.warpAffine(img, M_scaling, (new_width, new_height))
    return im_scaled


def resize_image(im, width, height):
    image1_resized = cv2.resize( im, (width, height))
    return image1_resized
```

```python
def position_image_on_background(background, image, position):
    # Cria uma cópia do background para trabalhar
    combined_image = background.copy()

    # Extrai as coordenadas da posição
    x_pos, y_pos = position

    # Posiciona a imagem no background conforme a posição fornecida
    combined_image[y_pos:y_pos+image.shape[0], x_pos:x_pos+image.shape[1]] = image

    return combined_image
```
